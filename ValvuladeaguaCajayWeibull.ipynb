```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import numpy as np
import matplotlib.pyplot as plt

# ========================================
# DATOS
# ========================================

datos = [
    1760,
    15180,
    2068,
    12584,
    8140,
    13948,
    27016,
    17226,
    8338,
    1980,
    5940,
    3762,
    10560,
    11814,
    4422,
    6534,
    2662,
    18128,
    5170,
    11440,
    11484
]

# ========================================
# GRÁFICO
# ========================================

fig, ax = plt.subplots(figsize=(10,6))

bp = ax.boxplot(
    datos,
    patch_artist=True,
    widths=0.35,
    showfliers=False,
    showmeans=True,
#    meanprops={
#        'marker': 'x',
#        'markeredgecolor': 'red',
#        'markersize': 10
#    }
)

bp['boxes'][0].set_facecolor('lightblue')

# ========================================
# Graficar los datos individuales como puntos
# ========================================

# Posición horizontal de los puntos (centrados en x=1)
x_puntos = np.random.normal(1, 0.02, size=len(datos))

ax.scatter(
    x_puntos,
    datos,
    color='blue',
    s=40,
    alpha=0.7,
    zorder=3,
    label='Datos'
)





# Estadísticos
minimo = np.min(datos)
q1 = np.percentile(datos, 25)
mediana = np.percentile(datos, 50)
media = np.mean(datos)
q3 = np.percentile(datos, 75)
maximo = np.max(datos)

# Posición horizontal de las etiquetas
x = 1.25

# Etiquetas
ax.text(x, minimo,
        f'Min = {minimo:.0f}',
        fontsize=10)

ax.text(x, q1,
        f'Q1 = {q1:.0f}',
        fontsize=10)

ax.text(x, mediana,
        f'Mediana = {mediana:.0f}',
        fontsize=10,
        color='blue',
        fontweight='bold')

ax.text(x, media,
        f'Media = {media:.0f}',
        fontsize=10,
        color='red',
        fontweight='bold')

ax.text(x, q3,
        f'Q3 = {q3:.0f}',
        fontsize=10)

ax.text(x, maximo,
        f'Max = {maximo:.0f}',
        fontsize=10)

# Identificación de outliers (criterio de Tukey)
IQR = q3 - q1
limite_inf = q1 - 1.5 * IQR
limite_sup = q3 + 1.5 * IQR

outliers = [valor for valor in datos
             if valor < limite_inf or valor > limite_sup]

# Etiquetar outliers
#for valor in outliers:
#    ax.text(
#        0.95,
#        valor,
#        str(int(valor)),
#        color='red',
#        fontsize=10
#    )

# Dar espacio para las etiquetas
ax.set_xlim(0.5, 1.6)

ax.set_title('Caja de Bigotes')
ax.set_ylabel('Horas de falla')
ax.set_xticks([])
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt
from reliability.Fitters import Fit_Weibull_2P

# ========================================
# DATOS
# ========================================
datos = [
    1760,
    15180,
    2068,
    12584,
    8140,
    13948,
    27016,
    17226,
    8338,
    1980,
    5940,
    3762,
    10560,
    11814,
    4422,
    6534,
    2662,
    18128,
    5170,
    11440,
    11484
]

# ========================================
# AJUSTE WEIBULL 2P
# ========================================
fit = Fit_Weibull_2P(
    failures=datos,
    show_probability_plot=False,
    print_results=False
)

beta = fit.beta
eta = fit.alpha

print(f"Beta (β) = {beta:.3f}")
print(f"Eta (η) = {eta:.1f} horas")

# ========================================
# CURVA DE CONFIABILIDAD
# ========================================
t = np.linspace(0, 30000, 1000)

R = np.exp(-(t/eta)**beta)

# Horas para confiabilidad 80% y 66%
R80 = 0.80
R70 = 0.70
R60 = 0.60

t80 = eta*(-np.log(R80))**(1/beta)
t70 = eta*(-np.log(R70))**(1/beta)
t60 = eta*(-np.log(R60))**(1/beta)

print(f'Horas para 80% de confiabilidad = {t80:.0f} h')
print(f'Horas para 70% de confiabilidad = {t70:.0f} h')
print(f'Horas para 60% de confiabilidad = {t60:.0f} h')

# ========================================
# GRAFICA
# ========================================
plt.figure(figsize=(8,6))

plt.plot(t, R*100, linewidth=2)

plt.axhline(80, color='green', linestyle='--')
plt.axhline(70, color='red', linestyle='--')
plt.axhline(60, color='blue', linestyle='--')

plt.axvline(t80, color='green', linestyle='--')
plt.axvline(t70, color='red', linestyle='--')
plt.axvline(t60, color='blue', linestyle='--')

plt.text(t80+500, 82,
         f'80% = {t80:.0f} h', color='green')

plt.text(t70+500, 72,
         f'70% = {t70:.0f} h', color='red')

plt.text(t60+500, 62,
         f'60% = {t60:.0f} h', color='blue')

plt.xlabel('Horas')
plt.ylabel('Confiabilidad (%)')
plt.title('Curva de Confiabilidad Weibull (2P)')
plt.grid()

plt.show()
```
